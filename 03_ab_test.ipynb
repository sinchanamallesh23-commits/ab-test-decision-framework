```python
import pandas as pd
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

df = pd.read_csv("data/clean/ab_clean.csv")
summary = df.groupby("group")["converted"].agg(users="size", conversions="sum", rate="mean")
summary["rate_%"] = (summary["rate"] * 100).round(3)
summary
```

```python
n_t, n_c = summary.loc["treatment", "users"], summary.loc["control", "users"]
x_t, x_c = summary.loc["treatment", "conversions"], summary.loc["control", "conversions"]
p_t, p_c = x_t / n_t, x_c / n_c

z, p_value = proportions_ztest([x_t, x_c], [n_t, n_c])

diff = p_t - p_c
se = np.sqrt(p_t * (1 - p_t) / n_t + p_c * (1 - p_c) / n_c)
ci_low, ci_high = diff - 1.96 * se, diff + 1.96 * se

print(f"Control:   {p_c*100:.3f}%")
print(f"Treatment: {p_t*100:.3f}%")
print(f"Difference: {diff*100:+.3f} percentage points ({diff/p_c*100:+.2f}% relative)")
print(f"95% CI for difference: {ci_low*100:+.3f} to {ci_high*100:+.3f} pp")
print(f"z = {z:.2f}, p-value = {p_value:.4f}")
```
