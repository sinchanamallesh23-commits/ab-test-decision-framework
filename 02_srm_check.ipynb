```python
import pandas as pd
from scipy.stats import chisquare

df = pd.read_csv("data/clean/ab_clean.csv")
counts = df["group"].value_counts()
stat, p = chisquare(counts, [len(df) / 2] * 2)

print(counts)
print("Treatment share:", round(counts["treatment"] / len(df) * 100, 3), "%")
print("SRM p-value:", round(p, 4))
print("SRM alarm!" if p < 0.001 else "No SRM: split is healthy")
```

```python
daily = df.groupby(["date", "group"]).size().unstack()
daily["treatment_share"] = (daily["treatment"] / daily.sum(axis=1) * 100).round(2)
daily["srm_p"] = daily.apply(lambda r: chisquare([r["control"], r["treatment"]])[1], axis=1).round(4)
daily
```
